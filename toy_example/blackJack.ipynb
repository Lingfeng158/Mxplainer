```python
def run_experiment(agent_control):
    states = []
    actions = []
    end = False
    valid_example = False
    while not valid_example:
        env = BlackJackEnv()
        state = env.state()
        valid_example = state[0][2]>11
    while not end:
        action = game_agent(state, agent_control)
        states.append(state[0])
        actions.append(action)
        state=env.step(action)
        end = state[1]
    return states, actions
```

```python
iterations_to_collect = 2000
control_param = opt_bj_agent

state_collector = []
action_collector = []
for i in range(iterations_to_collect):
    s, a = run_experiment(control_param)
    state_collector.extend(s)
    action_collector.extend(a)
```

```python
with open("bj_opt_data.json", "w") as f:
    json.dump({"state": state_collector, "action": action_collector}, f)
```
