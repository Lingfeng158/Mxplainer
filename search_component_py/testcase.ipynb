```python
def convert_pyDict_to_cMap(handDict):
    with open("tileInfo.txt","w") as f:
        f.write("{")
        for tile in handDict:
            if tile!="shown":
                f.write("{\"")
                f.write(tile)
                f.write("\",")
                f.write(str(handDict[tile]))
                f.write("}")
                f.write(",")
        f.write("};")
        f.write("\n")
        pack = handDict.get("shown", [])
        if pack!=[]:
            f.write("{")
            for entry in pack:
                f.write("{")
                for tile in entry:
                    f.write("{\"")
                    f.write(tile)
                    f.write("\",")
                    f.write(str(entry[tile]))
                    f.write("},")
                f.write("},")
            f.write("};")
        else:
            f.write("{};")
```

```python
# Bukao

def generateFullBuKao():
    handDictList = []
    typeList = ['B', "W", "T"]
    tileList = [[1,4,7],[2,5,8], [3,6,9]]
    for type1 in typeList:
        typeListReduced = typeList.copy()
        typeListReduced.remove(type1)
        for type2 in typeListReduced:
            if type2 == typeListReduced[0]:
                type3 = typeListReduced[1]
            else:
                type3 = typeListReduced[0]

            handDict = {}
            for rank in tileList[0]:
                handDict[type1+str(rank)]=1
            for rank in tileList[1]:
                handDict[type2+str(rank)]=1
            for rank in tileList[2]:
                handDict[type3+str(rank)]=1
            specialTileList = ["F1","F2","F3","F4","J1","J2","J3",]
            for tile in specialTileList:
                handDict[tile]=1
            key, val = random.choice(list(handDict.items()))
            handDict.pop(key)
            key, val = random.choice(list(handDict.items()))
            handDict.pop(key)
            handDictList.append(handDict)
    return handDictList

def generateFullMCDragon():
    handDictList = []
    typeList = ['B', "W", "T"]
    tileList = [[1,4,7],[2,5,8], [3,6,9]]
    for type1 in typeList:
        typeListReduced = typeList.copy()
        typeListReduced.remove(type1)
        for type2 in typeListReduced:
            if type2 == typeListReduced[0]:
                type3 = typeListReduced[1]
            else:
                type3 = typeListReduced[0]

            handDict = {}
            for rank in tileList[0]:
                handDict[type1+str(rank)]=1
            for rank in tileList[1]:
                handDict[type2+str(rank)]=1
            for rank in tileList[2]:
                handDict[type3+str(rank)]=1
            specialTileList = ["F1","F2","F3","F4","J1","J2","J3",]
            # key = random.choice(specialTileList)
            # handDict[key]=3
            # specialTileList.remove(key)
            key = random.choice(specialTileList)
            handDict[key]=2
            handDictList.append(handDict)
    return handDictList
```
